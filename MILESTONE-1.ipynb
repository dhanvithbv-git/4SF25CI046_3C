3. Data Transformation
   Perform the required transformations, such as changing data types, dropping unnecessary columns, or adding new columns.

```python
# Dropping unnecessary index columns
df = df.drop(columns=["Unnamed: 0.1", "Unnamed: 0"], errors="ignore")
print(df.head())
```

```python
# Changing data types
df["rate (out of 5)"] = pd.to_numeric(df["rate (out of 5)"], errors="coerce")
df["avg cost (two people)"] = pd.to_numeric(df["avg cost (two people)"], errors="coerce")
df["num of ratings"] = pd.to_numeric(df["num of ratings"], errors="coerce")
df["restaurant type"] = df["restaurant type"].astype("category")
df["online_order"] = df["online_order"].astype("category")
df["table booking"] = df["table booking"].astype("category")
print(df.dtypes)
```

```python
# Adding a new column: estimated cost per person
df["avg cost per person"] = df["avg cost (two people)"] / 2
print(df[["restaurant name", "avg cost (two people)", "avg cost per person"]].head())
```

4. Handling Null Values
   Identify null values in the dataset.
   Handle null values using mean imputation wherever applicable.

```python
# Checking null values
df.isnull().sum()
```

```python
# Filling numerical null values using mean
df["rate (out of 5)"] = df["rate (out of 5)"].fillna(df["rate (out of 5)"].mean())
df["avg cost (two people)"] = df["avg cost (two people)"].fillna(df["avg cost (two people)"].mean())
df["avg cost per person"] = df["avg cost per person"].fillna(df["avg cost per person"].mean())

# Filling missing categorical/text values using mode
for col in ["restaurant name", "restaurant type", "online_order", "table booking",
            "cuisines type", "area", "local address"]:
    if df[col].isnull().any() and not df[col].mode().empty:
        df[col] = df[col].fillna(df[col].mode()[0])

print("Null values after filling:")
df.isnull().sum()
```

```python
# Using apply and lambda
# Increasing restaurant ratings by 10% for demonstration
df["rating_after_10_percent"] = df["rate (out of 5)"].apply(lambda x: min(x * 1.10, 5))
print(df[["rate (out of 5)", "rating_after_10_percent"]].head())
```

5. Insight Generation
   Generate meaningful insights from the dataset using the concepts taught in class.
   Examples include sales trends over time and region-wise revenue analysis, depending on the selected dataset.


```python
# Grouping restaurants by area and finding average rating
area_rating = df.groupby("local address", observed=True)["rate (out of 5)"].mean().sort_values(ascending=False)
print(area_rating.head(10))
```

```python
# Grouping restaurants by restaurant type and finding average rating
type_rating = df.groupby("restaurant type", observed=True)["rate (out of 5)"].mean().sort_values(ascending=False)
print(type_rating)
```

```python
# Counting restaurants in each area
area_restaurants = df.groupby("local address", observed=True)["restaurant name"].count()
print(area_restaurants)
```

```python
# Sorting areas according to number of restaurants
area_restaurants = area_restaurants.sort_values(ascending=False)
print(area_restaurants.head(10))
```

```python
# Grouping by online order availability and finding average rating and cost
online_order_analysis = df.groupby("online_order", observed=True)[
    ["rate (out of 5)", "avg cost (two people)"]
].mean()
print(online_order_analysis)
```

```python
# Sorting restaurants according to rating
high_rated = df.sort_values("rate (out of 5)", ascending=False)
print(high_rated[["restaurant name", "local address", "rate (out of 5)"]].head(10))
```

```python
# Sorting restaurants according to average cost for two people
expensive_restaurants = df.sort_values("avg cost (two people)", ascending=False)
print(expensive_restaurants[["restaurant name", "local address", "avg cost (two people)"]].head(10))
```

```python
# Finding the most common cuisines
cuisine_count = df["cuisines type"].value_counts()
print(cuisine_count.head(10))
```

6. Data Visualization
   Visualize the generated insights using appropriate charts.


```python
# Getting top 10 areas by number of restaurants
top_areas = area_restaurants.head(10)
print(top_areas)
```

```python
# Bar graph: Top 10 areas by number of restaurants
top_areas.plot(kind="bar", figsize=(10, 5))
plt.xlabel("Area")
plt.ylabel("Number of Restaurants")
plt.title("Top 10 Areas by Number of Restaurants")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# Line graph: Average rating of top 10 areas
top_area_rating = area_rating.head(10)
plt.plot(top_area_rating.index.astype(str), top_area_rating.values, marker="o")
plt.xlabel("Area")
plt.ylabel("Average Rating")
plt.title("Average Rating of Top 10 Areas")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
